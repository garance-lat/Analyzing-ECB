```python
import requests
from bs4 import BeautifulSoup
from pathlib import Path

# URL of the ECB page that hosts the speeches dataset
url = "https://www.ecb.europa.eu/press/key/html/downloads.fr.html"

# 1️⃣ Download the page
response = requests.get(url)
response.raise_for_status()
html = response.text

# 2️⃣ Parse the HTML to find the CSV download link
soup = BeautifulSoup(html, "lxml")
csv_link = None
for a in soup.find_all("a", href=True):
    if "csv" in a["href"].lower():
        csv_link = a["href"]
        break

if not csv_link:
    raise ValueError("❌ No CSV link found on the ECB page.")

# 3️⃣ Build the absolute URL (the link is relative on the ECB website)
if not csv_link.startswith("http"):
    csv_link = "https://www.ecb.europa.eu" + csv_link

print("✅ Detected CSV link:", csv_link)

# 4️⃣ Download the CSV file
csv_response = requests.get(csv_link)
csv_response.raise_for_status()

# 5️⃣ Save locally
output_path = Path("ecb_speeches_dataset.csv")
with open(output_path, "wb") as f:
    f.write(csv_response.content)

print(f"💾 File downloaded and saved to: {output_path.resolve()}")
```
